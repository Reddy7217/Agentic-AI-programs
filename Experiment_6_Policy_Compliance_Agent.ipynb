# Experiment 6: Policy Compliance Agent

## Objective

To build a rule-based Policy Compliance Agent that evaluates
synthetic data against predefined policy rules and identifies
compliant and non-compliant records.

## Architecture

Synthetic Data
      ↓
Policy Compliance Agent
      ↓
Rule-Based Evaluation
      ↓
Compliance Decision
      ↓
Compliance Report

```python
print("Policy Compliance Agent")
print("Experiment 6")
```

```python
id="d9x4k2"
synthetic_data = [
    {
        "id": 1,
        "name": "Rahul",
        "age": 25,
        "amount": 5000,
        "category": "Standard"
    },
    {
        "id": 2,
        "name": "Priya",
        "age": 17,
        "amount": 3000,
        "category": "Standard"
    },
    {
        "id": 3,
        "name": "Arun",
        "age": 32,
        "amount": 15000,
        "category": "Standard"
    },
    {
        "id": 4,
        "name": "Sneha",
        "age": 28,
        "amount": 4000,
        "category": "Restricted"
    },
    {
        "id": 5,
        "name": "Vijay",
        "age": 40,
        "amount": 7000,
        "category": "Standard"
    }
]

print("Synthetic records created:", len(synthetic_data))

for record in synthetic_data:
    print(record)
```

```python
# Compliance rules

MINIMUM_AGE = 18
MAX_TRANSACTION_AMOUNT = 10000
RESTRICTED_CATEGORY = "Restricted"

print("Compliance rules created successfully!")
print("Minimum age:", MINIMUM_AGE)
print("Maximum transaction amount:", MAX_TRANSACTION_AMOUNT)
print("Restricted category:", RESTRICTED_CATEGORY)
```

```python
def compliance_agent(record):
    violations = []

    # Rule 1: Minimum age
    if record["age"] < MINIMUM_AGE:
        violations.append("Age is below the minimum requirement")

    # Rule 2: Maximum transaction amount
    if record["amount"] > MAX_TRANSACTION_AMOUNT:
        violations.append("Transaction amount exceeds the allowed limit")

    # Rule 3: Restricted category
    if record["category"] == RESTRICTED_CATEGORY:
        violations.append("Restricted category")

    # Final decision
    if len(violations) == 0:
        status = "COMPLIANT"
    else:
        status = "NON-COMPLIANT"

    return {
        "id": record["id"],
        "name": record["name"],
        "status": status,
        "violations": violations
    }


print("Policy Compliance Agent created successfully!")
```

```python
compliance_results = []

for record in synthetic_data:
    result = compliance_agent(record)
    compliance_results.append(result)

for result in compliance_results:
    print("Name:", result["name"])
    print("Status:", result["status"])
    print("Violations:", result["violations"])
    print("-" * 50)
```

```python
compliant_count = 0
non_compliant_count = 0

for result in compliance_results:
    if result["status"] == "COMPLIANT":
        compliant_count += 1
    else:
        non_compliant_count += 1

print("Compliance Report")
print("=" * 40)
print("Total Records:", len(compliance_results))
print("Compliant:", compliant_count)
print("Non-Compliant:", non_compliant_count)
```

```python
print("Detailed Compliance Report")
print("=" * 60)

for result in compliance_results:
    print("ID:", result["id"])
    print("Name:", result["name"])
    print("Status:", result["status"])

    if result["violations"]:
        print("Violations:")
        for violation in result["violations"]:
            print("-", violation)
    else:
        print("Violations: None")

    print("-" * 60)
```

```python
def check_compliance(record):
    result = compliance_agent(record)

    if result["status"] == "COMPLIANT":
        return f"{record['name']} is COMPLIANT with all policies."
    else:
        reasons = ", ".join(result["violations"])
        return f"{record['name']} is NON-COMPLIANT: {reasons}"


# Test the function
for record in synthetic_data:
    print(check_compliance(record))
```

```python
new_record = {
    "id": 6,
    "name": "Kiran",
    "age": 16,
    "amount": 20000,
    "category": "Restricted"
}

result = compliance_agent(new_record)

print("Compliance Test")
print("=" * 40)
print("Name:", result["name"])
print("Status:", result["status"])
print("Violations:")

for violation in result["violations"]:
    print("-", violation)
```
