# Experiment 5: Multi-Agent SDR System

## Objective

To design a multi-agent SDR system consisting of agents for
lead generation, lead qualification, and personalized email
generation.

## Architecture

Lead Generation Agent
        ↓
Lead Qualification Agent
        ↓
Email Generation Agent
        ↓
Personalized Email

```python
print("Multi-Agent SDR System")
print("Experiment 5")
```

```python
leads = [
    {
        "name": "Rahul",
        "company": "TechNova",
        "industry": "Software",
        "employees": 120,
        "email": "rahul@technova.com"
    },
    {
        "name": "Priya",
        "company": "HealthPlus",
        "industry": "Healthcare",
        "employees": 50,
        "email": "priya@healthplus.com"
    },
    {
        "name": "Arun",
        "company": "DataWorks",
        "industry": "Software",
        "employees": 300,
        "email": "arun@dataworks.com"
    },
    {
        "name": "Sneha",
        "company": "EduSmart",
        "industry": "Education",
        "employees": 30,
        "email": "sneha@edusmart.com"
    }
]

print("Number of leads:", len(leads))

for lead in leads:
    print(lead)
```

```python
def lead_generation_agent(leads):
    """
    Lead Generation Agent:
    Selects potential leads from the available lead data.
    """

    selected_leads = []

    for lead in leads:
        if lead["employees"] >= 50:
            selected_leads.append(lead)

    return selected_leads


generated_leads = lead_generation_agent(leads)

print("Generated Leads:")

for lead in generated_leads:
    print(lead)
```

```python
def lead_qualification_agent(leads):
    """
    Lead Qualification Agent:
    Checks whether a lead matches our target criteria.
    """

    qualified_leads = []

    for lead in leads:
        if lead["industry"] == "Software" and lead["employees"] >= 100:
            lead["qualified"] = True
            qualified_leads.append(lead)
        else:
            lead["qualified"] = False

    return qualified_leads


qualified_leads = lead_qualification_agent(generated_leads)

print("Qualified Leads:")

for lead in qualified_leads:
    print(lead)
```

```python
def email_generation_agent(lead):
    """
    Email Generation Agent:
    Creates a personalized sales email for a qualified lead.
    """

    email = f"""
Subject: AI Solutions for {lead["company"]}

Hi {lead["name"]},

I noticed that {lead["company"]} is a growing company in the
{lead["industry"]} industry.

We help companies use AI solutions to improve productivity,
automate repetitive tasks, and make better use of their data.

I would be happy to show you how these solutions could help
{lead["company"]}.

Would you be available for a short discussion?

Best regards,
SDR Team
"""

    return email


# Generate an email for the first qualified lead
email = email_generation_agent(qualified_leads[0])

print(email)
```

```python
def multi_agent_sdr(leads):
    # Agent 1: Generate potential leads
    generated_leads = lead_generation_agent(leads)

    # Agent 2: Qualify leads
    qualified_leads = lead_qualification_agent(generated_leads)

    # Agent 3: Generate emails
    results = []

    for lead in qualified_leads:
        email = email_generation_agent(lead)

        results.append({
            "lead": lead,
            "email": email
        })

    return results


# Run the complete SDR system
results = multi_agent_sdr(leads)

print("Multi-Agent SDR Results")
print("=" * 50)

for result in results:
    print("Lead:", result["lead"]["name"])
    print("Company:", result["lead"]["company"])
    print("Email:")
    print(result["email"])
    print("=" * 50)
```

```python
new_leads = [
    {
        "name": "Kiran",
        "company": "CloudTech",
        "industry": "Software",
        "employees": 200,
        "email": "kiran@cloudtech.com"
    },
    {
        "name": "Anita",
        "company": "RetailPro",
        "industry": "Retail",
        "employees": 80,
        "email": "anita@retailpro.com"
    }
]

results = multi_agent_sdr(new_leads)

for result in results:
    print("Qualified Lead:", result["lead"]["name"])
    print("Company:", result["lead"]["company"])
    print(result["email"])
    print("-" * 60)
```

```python
def lead_status_report(leads):
    generated_leads = lead_generation_agent(leads)
    qualified_leads = lead_qualification_agent(generated_leads)

    qualified_names = [lead["name"] for lead in qualified_leads]

    print("Lead Status Report")
    print("=" * 40)

    for lead in generated_leads:
        if lead["name"] in qualified_names:
            print(f"{lead['name']} - QUALIFIED")
        else:
            print(f"{lead['name']} - NOT QUALIFIED")


lead_status_report(new_leads)
```

# Final Result

The Multi-Agent SDR System was successfully implemented.

The system consists of three agents:

1. Lead Generation Agent
   - Identifies potential leads from the available lead data.

2. Lead Qualification Agent
   - Evaluates leads based on predefined criteria.

3. Email Generation Agent
   - Creates personalized sales emails for qualified leads.

## Complete Workflow

Lead Data
    ↓
Lead Generation Agent
    ↓
Potential Leads
    ↓
Lead Qualification Agent
    ↓
Qualified Leads
    ↓
Email Generation Agent
    ↓
Personalized Sales Email

## Conclusion

The experiment demonstrates how multiple specialized agents
can work together as a single workflow for an SDR process.
