```python
from google.colab import files

uploaded = files.upload()

image_path = list(uploaded.keys())[0]

print("Image uploaded:", image_path)
```

```python
from PIL import Image
import matplotlib.pyplot as plt

image = Image.open(image_path)

plt.imshow(image)
plt.axis("off")
plt.show()
```

```python
from transformers import BlipProcessor, BlipForQuestionAnswering

processor = BlipProcessor.from_pretrained(
    "Salesforce/blip-vqa-base"
)

model = BlipForQuestionAnswering.from_pretrained(
    "Salesforce/blip-vqa-base"
)

print("Visual QA model loaded successfully!")
```

```python
question = "What is in this image?"

inputs = processor(
    image,
    question,
    return_tensors="pt"
)

outputs = model.generate(
    **inputs,
    max_new_tokens=30
)

answer = processor.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Question:", question)
print("Answer:", answer)
```

```python
question = "What color is the object?"

inputs = processor(
    image,
    question,
    return_tensors="pt"
)

outputs = model.generate(
    **inputs,
    max_new_tokens=30
)

answer = processor.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Question:", question)
print("Answer:", answer)
```

```python
def ask_question(image, question):
    inputs = processor(
        image,
        question,
        return_tensors="pt"
    )

    outputs = model.generate(
        **inputs,
        max_new_tokens=30
    )

    answer = processor.decode(
        outputs[0],
        skip_special_tokens=True
    )

    print("Question:", question)
    print("Answer:", answer)
```

```python
ask_question(image, "What is in this image?")
```

```python
ask_question(image, "How many objects are visible?")
```

```python
ask_question(image, "What color is the main object?")
```

```python
ask_question(image, "Where is the object?")
```
