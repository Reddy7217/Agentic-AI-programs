# Experiment 9: Reasoning Model Benchmarking

## Objective

To compare the outputs of a reasoning model using different
prompting strategies and evaluate their effectiveness on
reasoning tasks.

## Prompting Strategies

1. Direct Prompt
2. Step-by-Step Prompt
3. Structured Prompt

## Workflow

Reasoning Problem
      ↓
Different Prompting Strategies
      ↓
Model Outputs
      ↓
Output Comparison
      ↓
Benchmark Results

```python
print("Reasoning Model Benchmarking")
print("Experiment 9")
```

```python
reasoning_problem = """
A store has 5 boxes.
Each box contains 12 pencils.
The store sells 18 pencils.
Then it receives 24 more pencils.

How many pencils does the store have now?
"""

print("Benchmark Problem:")
print(reasoning_problem)
```

```python
direct_prompt = f"""
{reasoning_problem}

Give the final answer.
"""

print("Direct Prompt:")
print(direct_prompt)
```

```python
step_by_step_prompt = f"""
Solve the following problem step by step.

Show the calculation at each step, then provide the final answer.

Problem:
{reasoning_problem}
"""

print("Step-by-Step Prompt:")
print(step_by_step_prompt)
```

```python
structured_prompt = f"""
Solve the following reasoning problem using this structure:

1. Identify the important information.
2. Perform the calculations.
3. Verify the result.
4. Give the final answer clearly.

Problem:
{reasoning_problem}
"""

print("Structured Prompt:")
print(structured_prompt)
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

print("Reasoning model loaded successfully!")
```

```python
def run_model(prompt):
    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True
    )

    outputs = model.generate(
        **inputs,
        max_new_tokens=100
    )

    answer = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return answer

print("Model function created successfully!")
```

```python
direct_output = run_model(direct_prompt)

print("Direct Prompt Output:")
print(direct_output)
```

```python
step_by_step_output = run_model(step_by_step_prompt)

print("Step-by-Step Prompt Output:")
print(step_by_step_output)
```

```python
structured_output = run_model(structured_prompt)

print("Structured Prompt Output:")
print(structured_output)
```

```python
print("=" * 60)
print("REASONING MODEL BENCHMARK RESULTS")
print("=" * 60)

print("\n1. DIRECT PROMPT")
print(direct_output)

print("\n2. STEP-BY-STEP PROMPT")
print(step_by_step_output)

print("\n3. STRUCTURED PROMPT")
print(structured_output)
```

```python
benchmark_results = {
    "Prompt Strategy": [
        "Direct Prompt",
        "Step-by-Step Prompt",
        "Structured Prompt"
    ],
    "Model Output": [
        direct_output,
        step_by_step_output,
        structured_output
    ]
}

import pandas as pd

benchmark_df = pd.DataFrame(benchmark_results)

display(benchmark_df)
```

```python
id="expected_answer"
expected_answer = 66

print("Expected Answer:", expected_answer)
```

```python
def check_answer(output, expected_answer):
    return str(expected_answer) in output

evaluation_results = {
    "Prompt Strategy": [
        "Direct Prompt",
        "Step-by-Step Prompt",
        "Structured Prompt"
    ],
    "Correct Answer Found": [
        check_answer(direct_output, expected_answer),
        check_answer(step_by_step_output, expected_answer),
        check_answer(structured_output, expected_answer)
    ]
}

evaluation_df = pd.DataFrame(evaluation_results)

display(evaluation_df)
```

```python
print("=" * 60)
print("BENCHMARK SUMMARY")
print("=" * 60)

for strategy, correct in zip(
    evaluation_results["Prompt Strategy"],
    evaluation_results["Correct Answer Found"]
):
    status = "Correct" if correct else "Incorrect"
    print(f"{strategy}: {status}")
```

```python
accuracy = (
    sum(evaluation_results["Correct Answer Found"])
    / len(evaluation_results["Correct Answer Found"])
) * 100

print(f"Overall Benchmark Accuracy: {accuracy:.2f}%")
```
