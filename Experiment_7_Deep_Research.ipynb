# Experiment 7: Deep Research Agent Workflow

## Objective

To implement a Deep Research Agent workflow that uses planning
and reflection to improve content generation.

## Architecture

Research Topic
      ↓
Planning Agent
      ↓
Research Questions
      ↓
Information Gathering
      ↓
Reflection Agent
      ↓
Improved Content
      ↓
Final Content

```python
print("Deep Research Agent Workflow")
print("Experiment 7")
```

```python
research_topic = "Artificial Intelligence in Healthcare"

print("Research Topic:", research_topic)
```

```python
def planning_agent(topic):
    """
    Planning Agent:
    Breaks a research topic into important research questions.
    """

    research_questions = [
        f"What is {topic}?",
        f"What are the major applications of {topic}?",
        f"What are the benefits of {topic}?",
        f"What are the challenges and limitations of {topic}?",
        f"What are the future developments of {topic}?"
    ]

    return research_questions


research_questions = planning_agent(research_topic)

print("Research Plan")
print("=" * 50)

for i, question in enumerate(research_questions, 1):
    print(f"{i}. {question}")
```

```python
def research_agent(questions):
    """
    Research Agent:
    Collects information for each research question.
    In this experiment, the information is represented
    using synthetic research data.
    """

    research_results = {
        questions[0]: """
        Artificial Intelligence in Healthcare refers to the use of
        AI techniques such as machine learning and data analysis
        to support healthcare-related tasks and decision making.
        """,

        questions[1]: """
        Major applications include medical image analysis,
        disease prediction, patient monitoring, drug discovery,
        and healthcare administration.
        """,

        questions[2]: """
        AI can help process large amounts of healthcare data,
        automate repetitive tasks, support clinical decisions,
        and enable more personalized healthcare.
        """,

        questions[3]: """
        Important challenges include data privacy, biased or
        incomplete data, reliability, human oversight, and
        integration with existing healthcare systems.
        """,

        questions[4]: """
        Future developments may include improved clinical
        decision-support systems, personalized healthcare,
        AI-assisted medical research, and greater automation.
        """
    }

    return research_results


research_results = research_agent(research_questions)

print("Research Results")
print("=" * 60)

for question, answer in research_results.items():
    print("\nQuestion:", question)
    print("Research:", answer.strip())
```

```python
def reflection_agent(questions, research_results):
    """
    Reflection Agent:
    Checks whether every research question has
    sufficient research information.
    """

    gaps = []

    for question in questions:
        if question not in research_results:
            gaps.append(question)
        elif not research_results[question].strip():
            gaps.append(question)

    if len(gaps) == 0:
        reflection = "Research is complete. No major gaps were found."
    else:
        reflection = "Additional research is required."

    return reflection, gaps


reflection, gaps = reflection_agent(
    research_questions,
    research_results
)

print("Reflection Result")
print("=" * 50)
print(reflection)

if gaps:
    print("\nResearch Gaps:")
    for gap in gaps:
        print("-", gap)
else:
    print("\nNo research gaps found.")
```

```python
def content_generation_agent(topic, research_results):
    """
    Content Generation Agent:
    Uses the research results to create structured content.
    """

    definition = research_results[
        f"What is {topic}?"
    ]

    applications = research_results[
        f"What are the major applications of {topic}?"
    ]

    benefits = research_results[
        f"What are the benefits of {topic}?"
    ]

    challenges = research_results[
        f"What are the challenges and limitations of {topic}?"
    ]

    future = research_results[
        f"What are the future developments of {topic}?"
    ]

    content = f"""
# {topic}

## Introduction

{definition.strip()}

## Major Applications

{applications.strip()}

## Benefits

{benefits.strip()}

## Challenges and Limitations

{challenges.strip()}

## Future Developments

{future.strip()}

## Conclusion

{topic} has the potential to support healthcare by improving
data analysis, decision support, research, and automation.
However, responsible implementation requires attention to
privacy, data quality, human oversight, and system reliability.
"""

    return content


draft_content = content_generation_agent(
    research_topic,
    research_results
)

print(draft_content)
```

```python
def content_reflection_agent(content):
    """
    Reflection Agent:
    Checks whether the generated content contains
    all required research sections.
    """

    required_sections = [
        "Introduction",
        "Major Applications",
        "Benefits",
        "Challenges and Limitations",
        "Future Developments",
        "Conclusion"
    ]

    missing_sections = []

    for section in required_sections:
        if section not in content:
            missing_sections.append(section)

    if len(missing_sections) == 0:
        reflection = "Draft content is complete. All required sections are present."
    else:
        reflection = "Draft content needs improvement."

    return reflection, missing_sections


draft_reflection, missing_sections = content_reflection_agent(
    draft_content
)

print("Content Reflection")
print("=" * 50)
print(draft_reflection)

if missing_sections:
    print("\nMissing Sections:")
    for section in missing_sections:
        print("-", section)
else:
    print("\nNo missing sections found.")
```

```python
def deep_research_workflow(topic):
    # Step 1: Planning
    questions = planning_agent(topic)

    # Step 2: Research
    research = research_agent(questions)

    # Step 3: Reflection on research
    reflection, gaps = reflection_agent(questions, research)

    # Step 4: Generate content
    draft = content_generation_agent(topic, research)

    # Step 5: Reflect on generated content
    content_check, missing_sections = content_reflection_agent(draft)

    return {
        "questions": questions,
        "research": research,
        "research_reflection": reflection,
        "research_gaps": gaps,
        "draft": draft,
        "content_reflection": content_check,
        "missing_sections": missing_sections
    }


final_result = deep_research_workflow(research_topic)

print("Deep Research Workflow Completed!")
print("=" * 60)
print("Research Reflection:", final_result["research_reflection"])
print("Content Reflection:", final_result["content_reflection"])
```

```python
print("FINAL RESEARCH CONTENT")
print("=" * 70)

print(final_result["draft"])
```

# Final Result

The Deep Research Agent Workflow was successfully implemented.

## Main Components

1. Planning Agent
   - Breaks the research topic into smaller questions.

2. Research Agent
   - Collects information for each research question.

3. Reflection Agent
   - Checks the research for missing information.

4. Content Generation Agent
   - Creates structured content from the research.

5. Content Reflection Agent
   - Reviews the generated content for missing sections.

## Complete Workflow

Research Topic
      ↓
Planning
      ↓
Research
      ↓
Reflection
      ↓
Content Generation
      ↓
Content Reflection
      ↓
Final Content

be combined with content generation to create a structured
Deep Research Agent workflow.
